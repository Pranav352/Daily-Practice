```python
def reverse_string(s):
    stack = []
    for char in s:
        stack.append(char)
    rstr = ""
    while stack:
        rstr += stack.pop()
    return rstr

if __name__ == "__main__":
    print(reverse_string("We will conquere COVID-19"))
```
