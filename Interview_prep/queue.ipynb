```python
from collections import deque

class Queue:
    
    def __init__(self):
        self.buffer = deque()
    
    def enqueue(self, val):
        self.buffer.appendleft(val)
        
    def dequeue(self):
        return self.buffer.pop()
    
    def is_empty(self):
        return len(self.buffer)==0
    
    def size(self):
        return len(self.buffer)
```

```python
pq = Queue()

pq.enqueue({
    'company': 'amezon',
    'timestamp': '27 apr, 11.01 AM',
    'price': 121.10
})
pq.enqueue({
    'company': ' amezon',
    'timestamp': '27 apr, 11.02 AM',
    'price': 131
})
pq.enqueue({
    'company': 'amezon',
    'timestamp': '27 apr, 11.03 AM',
    'price': 141
})

```
